```python
import pandas as pd
url ="https://raw.githubusercontent.com/kflisikowsky/AWDP/refs/heads/main/data/dane.csv"
zbior=pd.read_csv(url)
zbior.head(5)
```

```python
def convert_total_sqft(value):
    """Convert a total_sqft value to a float.

    For ranges such as '2100 - 2850', the midpoint is returned.
    Unsupported or missing values return None.
    """
    if pd.isna(value):
        return None

    value = str(value).strip()
    if not value:
        return None

    try:
        return float(value)
    except ValueError:
        parts = [part.strip() for part in value.split("-")]
        if len(parts) == 2:
            try:
                return (float(parts[0]) + float(parts[1])) / 2
            except ValueError:
                return None
        return None


zbior["total_sqft"] = zbior["total_sqft"].apply(convert_total_sqft)
```
